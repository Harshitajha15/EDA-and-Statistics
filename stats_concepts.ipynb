```python
frequency={}
for i in data:
    count=0
    for j in data:
        if i==j:
            count=count+1
    # print(f"no. of {i} is equal to {count}")
    fequen={i:count}
    # print(fequen)
    frequency.update(fequen)
print(frequency)

```

```python
np.var(data)
```

```python
def vari(data):
    sum=0
    # count=0
    len(data)
    for i in data:
        sum=sum+i
        print(sum)
        # count=count+1
    # print("hhhhhhhhhhhh",count)
    mean=sum/len(data)
    sum1=0

    for i in data:
        a=i-mean
        b=a*a
        sum1=sum1+b
    
    ans=sum1/len(data)
    return ans

resul=vari(data)
print(resul)


```
